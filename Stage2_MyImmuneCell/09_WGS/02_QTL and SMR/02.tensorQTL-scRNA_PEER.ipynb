```r
# 设置目标文件夹路径
folder_path <- "/media/AnalysisDisk2/Yangshichen/2_My-Onek/WGS/Data/scRNA-seq/04.top2000_normal_dis/"

# 获取所有 csv 文件的完整路径
csv_files <- list.files(path = folder_path, pattern = "\\.csv$", full.names = TRUE)
csv_files
```

```r
library(parallel)
num_cores <- 60
```

```r
process_file <- function(file) {
  filename <- basename(file)
  celltype <- tools::file_path_sans_ext(filename)
  
  expr <- read.csv(file, header = TRUE)
  colnames(expr)[1] <- 'sample'
  rownames(expr) <- expr[, 1]
  expr <- expr[, -1]
  
  # 样本匹配
  meta <- meatdata[match(rownames(expr), rownames(meatdata)), ]
  meta$Gender <- ifelse(meta$Gender == "Male", 1, 0)
  meta$Batch <- as.numeric(gsub("B", "", meta$Batch))
  meta$Pregnancy <- ifelse(meta$Pregnancy == "YES", 1, 0)
  #meta$Plate.ID <- as.numeric(factor(meta$Plate.ID))
  meta <- as.data.frame(lapply(meta, function(x) as.numeric(as.character(x))))
  
  # 设置 PEER 参数
  model <- PEER()
  PEER_setPhenoMean(model, as.matrix(expr))
  PEER_setCovariates(model, as.matrix(meta))
  dim(PEER_getPhenoMean(model))
  
  # 设置生成 60 个 PEER 因子：K=60
  K <- 60
  PEER_setNk(model, K)
  PEER_getNk(model)
  PEER_setNmax_iterations(model, 1000)
  
  # 计算
  PEER_update(model)
  
  # 输出结果
  pf.names <- paste('PF', 1:K, sep = '')
  
  # factors: NxK 矩阵
  f <- PEER_getX(model)
  f_df <- data.frame(f)
  colnames(f_df) <- c(colnames(meatdata), pf.names)
  rownames(f_df) <- rownames(expr)
  write.csv(f_df, paste0("/media/AnalysisDisk2/Yangshichen/2_My-Onek/WGS/Data/scRNA-seq/05.PEER/01.factor/", celltype, ".csv"))
  
  # weights: GxK 矩阵
  w <- PEER_getW(model)
  w_df <- data.frame(w)
  colnames(w_df) <- c(colnames(meatdata), pf.names)
  w_df$geneid <- colnames(expr)
  rownames(w_df) <- w_df$geneid
  w_df <- w_df[, -c(length(colnames(w_df)))]
  write.csv(w_df, paste0("/media/AnalysisDisk2/Yangshichen/2_My-Onek/WGS/Data/scRNA-seq/05.PEER/02.weights/", celltype, ".csv"))
  
  # precision: Kx1 矩阵
  p <- PEER_getAlpha(model)
  p_df <- data.frame(p)
  rownames(p_df) <- c(colnames(meatdata), pf.names)
  write.csv(p_df, paste0("/media/AnalysisDisk2/Yangshichen/2_My-Onek/WGS/Data/scRNA-seq/05.PEER/03.precision/", celltype, ".csv"))
  
  # residuals: NxG 矩阵
  r <- PEER_getResiduals(model)
  r_df <- data.frame(r)
  colnames(r_df) <- colnames(expr)
  rownames(r_df) <- rownames(expr)
  write.csv(r_df, paste0("/media/AnalysisDisk2/Yangshichen/2_My-Onek/WGS/Data/scRNA-seq/05.PEER/04.residuals/", celltype, ".csv"))
  
  print(paste0(celltype, " calculate over!"))
}
```

```r
mclapply(csv_files, process_file, mc.cores = num_cores)
```

```r
for (file in csv_files) {
    filename <- basename(file)
    celltype <- tools::file_path_sans_ext(filename)
    
    expr <- read.csv(file,header = T)
    colnames(expr)[1] <- 'sample'
    rownames(expr) <- expr[,1]
    expr <- expr[,-1]

    #样本匹配
    meta <- meatdata[match(rownames(expr), rownames(meatdata)), ]
    meta$Gender <- ifelse(meta$Gender == "Male", 1, 0)
    meta$Batch <- as.numeric(gsub("B", "", meta$Batch))
    meta$Pregnancy <- ifelse(meta$Pregnancy == "YES", 1, 0)
    meta <- as.data.frame(lapply(meta, function(x) as.numeric(as.character(x))))
    
    #Set PEER paramaters based on the instructions from PEER package website
    model = PEER()
    PEER_setPhenoMean(model, as.matrix(expr))
    PEER_setCovariates(model, as.matrix(meta))
    dim(PEER_getPhenoMean(model))

    #Set to generate 60 PEER factors: K=60
    K <- 60
    PEER_setNk(model,K)
    PEER_getNk(model)

    PEER_setNmax_iterations(model, 1000)

    #Calculate
    PEER_update(model)

    #Observing output
    pf.names <- paste('PF',1:K, sep = '')
    
    # factors: NxK matrix
    f <- PEER_getX(model)
    f_df <- data.frame(f)
    colnames(f_df) <- c(colnames(meatdata),pf.names)
    rownames(f_df) <- rownames(expr)
    write.csv(f_df, paste0("/media/AnalysisDisk2/Yangshichen/2_My-Onek/WES/Germline/Data/scRNA-seq/05.PEER/01.factor/",celltype,".csv"))

    # weights: GxK matrix
    w <- PEER_getW(model)
    w_df <- data.frame(w)
    colnames(w_df) <- c(colnames(meatdata),pf.names)
    w_df$geneid <- colnames(expr)
    rownames(w_df) <- w_df$geneid
    w_df <- w_df[,-c(length(colnames(w_df)))]
    write.csv(w_df, paste0("/media/AnalysisDisk2/Yangshichen/2_My-Onek/WES/Germline/Data/scRNA-seq/05.PEER/02.weights/",celltype,".csv"))
    
    # precision: Kx1 matrix
    p <- PEER_getAlpha(model)
    p_df <- data.frame(p)
    rownames(p_df) <- c(colnames(meatdata),pf.names)
    write.csv(p_df, paste0("/media/AnalysisDisk2/Yangshichen/2_My-Onek/WES/Germline/Data/scRNA-seq/05.PEER/03.precision/",celltype,".csv"))

    # residuals: NxG matrix
    r <- PEER_getResiduals(model)
    r_df <- data.frame(r)
    colnames(r_df) <- colnames(expr)
    rownames(r_df) <- rownames(expr)
    write.csv(r_df, paste0("/media/AnalysisDisk2/Yangshichen/2_My-Onek/WES/Germline/Data/scRNA-seq/05.PEER/04.residuals/",celltype,".csv"))
    print(paste0(celltype," calculate over!"))
}
```
