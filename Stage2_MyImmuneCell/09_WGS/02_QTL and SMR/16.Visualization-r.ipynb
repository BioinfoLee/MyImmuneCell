```r
top_snp <- "1:160902873:G:A"
merged$is_top <- ifelse(merged$snp == top_snp, "Top SNP", "Others")
```

```r
p1 <- ggplot(merged, aes(x = bp_eQTL, y = logP_eQTL)) +
  geom_point(aes(color = is_top), size = 2.5, alpha = 0.8) +
  geom_vline(xintercept = merged$bp[merged$snp == top_snp],
             linetype = "dashed", color = "black") +
  geom_text_repel(
    data = subset(merged, is_top == "Top SNP"),
    aes(label = snp),
    color = "red", size = 3
  ) +
  scale_color_manual(values = c("Others" = "steelblue", "Top SNP" = "red")) +
  labs(
    title = "region (eQTL)",
    x = "Genomic position (bp)",
    y = expression(-log[10](italic(P))),
    color = ""
  ) +
  theme_bw() +
  theme(plot.title = element_text(hjust = 0.5, face = "bold"),
        legend.position = "top")
ggsave('/media/AnalysisDisk2/Yangshichen/2_My-Onek/WES/Germline/Result/SMR_plot/coloc_plot/CD244_eqtl-CD56dim NK cells.pdf',p1,width=5,height=3)
```

```r
data$group <- ifelse(grepl("NDN|LDN|Basophils", data$celltype), "DN", "nonDN")
data
```

```r
gene_group <- data %>%
  group_by(Gene) %>%
  summarise(groups = list(unique(group)))
```

```r
DN_specific <- gene_group %>%
  filter(
    sapply(groups, function(x) {
      all(x %in% "DN")  # 只出现在 DN
    })
  ) %>%
  pull(Gene)
DN_specific
```

```r
calc <- function(feature){

  stat <- data %>%
    group_by(.data[[feature]]) %>%
    summarise(group_set = paste(unique(group), collapse = ","))

  stat$category <- ifelse(
    stat$group_set == "DN", "DN-specific",
    ifelse(stat$group_set == "nonDN", "nonDN-specific", "Shared")
  )

  stat %>%
    group_by(category) %>%
    summarise(count = n())
}
```

```r
snp_df <- calc("topSNP")
snp_df
p <- ggplot(snp_df, aes(x = "", y = count, fill = category)) +
  geom_bar(stat = "identity", width = 1) +
  coord_polar(theta = "y") +
  theme_void() +
  labs(title = "Gene distribution (DN vs nonDN)")
p
ggsave('/media/AnalysisDisk2/Yangshichen/2_My-Onek/WGS/Result/plot/SMR/summary_snp_celltype.pdf',p, width=10)
```

```r
snp_df <- calc("Gene")
snp_df
p <- ggplot(snp_df, aes(x = "", y = count, fill = category)) +
  geom_bar(stat = "identity", width = 1) +
  coord_polar(theta = "y") +
  theme_void() +
  labs(title = "Gene distribution (DN vs nonDN)")
p
ggsave('/media/AnalysisDisk2/Yangshichen/2_My-Onek/WGS/Result/plot/SMR/summary_gene_celltype.pdf',p, width=10)
```
