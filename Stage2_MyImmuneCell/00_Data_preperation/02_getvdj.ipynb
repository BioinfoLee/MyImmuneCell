### TCR

```r
#shell MyImmuCell/01_rawdata/tcr.sh
sample=$1
sample_scRNA=$2
/home/liyanguo/anaconda3/envs/celescope/bin/celescope vdj mapping_vdj --outdir ./MyImmuCell_TCR_out/${sample}/04.mapping_vdj --sample ${sample} --thread 16 \
--ref_path /home/liyanguo/MyImmuCell/01_rawdata/Human_vdj/hs_TR --type TCR --species human \
--fasta /localdisk/immune/0_raw_immune1k/celescope_v2/MyImmuCell_TCR/${sample}_${sample}_consensus.fasta
/home/liyanguo/anaconda3/envs/celescope/bin/celescope vdj count_vdj --outdir ./MyImmuCell_TCR_out/${sample}/05.count_vdj --sample ${sample} --thread 16 \
--type TCR --UMI_min 3 --BCR_iUMI 2 --TCR_iUMI 1 --expected_target_cell_num 10000 --target_weight 3.0  \
--match_dir ./MyImmuCell_scRNA/${sample_scRNA} --UMI_count_filter_file ./MyImmuCell_TCR_out/${sample}/04.mapping_vdj/${sample}_UMI_count_filtered.tsv
```

```r
#run parallel
cd /home/liyanguo/MyImmuCell/01_rawdata
my_func() {
    sh tcr.sh $1 $2
}
export -f my_func
nohup cat tcrlist | nohup parallel -j 5 my_func &
```

### BCR
