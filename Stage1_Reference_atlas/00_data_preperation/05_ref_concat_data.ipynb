```python
import scanpy as sc
import numpy as np 
import pandas as pd 
import anndata as ad
from glob import glob
from tqdm import tqdm
```

# 1. Concat PBMC scRNA data

```python
files = glob('/home/liyanguo/MyImmuCell/02_Read_QC/Ref_RNA_h5ad/*')
files
```

```python
len(files)
```

```python
merged_chunk={}
chunk_size = 200
```

```python
for chunk_start in range(0,len(files),chunk_size):
    data_dic = {}
    chunk_files = files[chunk_start:chunk_start+chunk_size]

    for i in tqdm(chunk_files):
        sample_name = i.split('/')[-1].split('.')[0]
        adata = sc.read_h5ad(i)
        data_dic[sample_name] = adata
    chunk_adatas = sc.concat(data_dic,join='outer')
    merged_chunk[chunk_start] = chunk_adatas
```

```python
adata = sc.concat(merged_chunk,join='outer')
```
