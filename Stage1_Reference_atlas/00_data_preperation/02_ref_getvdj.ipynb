### TCR

```r
#shell MyImmuCell/01_rawdata/ref_tcr.sh
sample=$1
sample_scRNA=$2
/home/liyanguo/anaconda3/envs/celescope/bin/celescope vdj mapping_vdj --outdir ./Reference_Atlas_TCR_out/${sample}/04.mapping_vdj --sample ${sample} --thread 16 \
--ref_path /home/liyanguo/MyImmuCell/01_rawdata/Human_vdj/hs_TR --type TCR --species human \
--fasta /localdisk/immune/0_raw_immune1k/celescope_v2/Reference_Atlas_TCR/${sample}_${sample}_consensus.fasta
/home/liyanguo/anaconda3/envs/celescope/bin/celescope vdj count_vdj --outdir ./Reference_Atlas_TCR_out/${sample}/05.count_vdj --sample ${sample} --thread 16 \
--type TCR --UMI_min 3 --BCR_iUMI 2 --TCR_iUMI 1 --expected_target_cell_num 10000 --target_weight 3.0  \
--match_dir ./Reference_Atlas_scRNA/${sample_scRNA} --UMI_count_filter_file ./Reference_Atlas_TCR_out/${sample}/04.mapping_vdj/${sample}_UMI_count_filtered.tsv
```

```r
#example
sh ref_tcr.sh CRS20230712_N1-T CRS20230712_N1_EmptyDrops_CR_310270_matrix
```

```r
#run parallel
cd /home/liyanguo/MyImmuCell/01_rawdata
my_func() {
    sh ref_tcr.sh $1 $2
}
export -f my_func
nohup cat tcrlist.txt | nohup parallel -j 3 my_func &
```

### BCR

```r
#shell MyImmuCell/01_rawdata/ref_bcr.sh
sample=$1
sample_scRNA=$2
/home/liyanguo/anaconda3/envs/celescope/bin/celescope vdj mapping_vdj --outdir ./Reference_Atlas_BCR_out/${sample}/04.mapping_vdj --sample ${sample} --thread 16 \
--ref_path /home/liyanguo/MyImmuCell/01_rawdata/Human_vdj/hs_IG --type BCR --species human  \
--fasta /localdisk/immune/0_raw_immune1k/celescope_v2/Reference_Atlas_BCR/${sample}_${sample}_consensus.fasta
/home/liyanguo/anaconda3/envs/celescope/bin/celescope vdj count_vdj --outdir ./Reference_Atlas_BCR_out/${sample}/05.count_vdj --sample ${sample} --thread 16 \
--type BCR --UMI_min 3 --BCR_iUMI 2 --TCR_iUMI 1 --expected_target_cell_num 3000 --target_weight 3.0  \
--match_dir ./Reference_Atlas_scRNA/${sample_scRNA} --UMI_count_filter_file ./Reference_Atlas_BCR_out/${sample}/04.mapping_vdj/${sample}_UMI_count_filtered.tsv
```
